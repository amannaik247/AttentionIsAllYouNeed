### forward method in residual connection:
```
def forward(self, x, sublayer):
  return x + self.dropout(sublayer(self.norm(x)))
```

- once inside the encoder block, the input embeds go into multihead attention and also directly get addded after normalisation
- BUT in code we normalise the embeds first(pre norm[due to advice from other papers]) then pass it to multihead, add dropout and finally add it to the original embeds
